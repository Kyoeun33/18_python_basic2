```python
# # 덧셈함수 만들기
# def 함수이름(매개변수1,매개변수2):
#     매개변수를 사용한 코드

def add(a,b):
    print(a+b)

result1=add(10,20)
print(result1)   ## 반환 받지 못함
```

```python
# 함수의 결과를 반환하기
# def 함수이름(매개변수1,매개변수2):
#       return 매개 변수를 이용한 코드

def add(a,b):
    return a+b

resut2=add(10,20)
print(resut2)  ##반환 받음
```

```python
# return 으로 함수 중간에서 빠져나오기

def not_ten(a):
    if a == 10:
        return
    print(a,"입니다. ",sep="")
    
```

```python
not_ten(3)
```

```python
not_ten(10)
```

```python
# 함수에서 값을 여러개 반환하기
def add_sub(a,b):
    return a+b
    return a-b

add_sub(10,20)
```

```python
def add_sub(a,b):
    return a+b , a-b 

result = add_sub(10,20)
result
```

```python
# 함수의 호출 과정 알아보기

def mul(a,b):
    c = a*b
    return c

def add(a,b):
    c = a+b
    print(c)
    d = mul(a,b)
    print(d)

x= 10
y= 20
add(x,y)
```
