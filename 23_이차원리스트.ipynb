```python
#심사문제

col, row = map(int,input().split())
matrix=[]
for _ in range(row):
    matrix.append(list(input()))

for i in range(row) :
    for j in range(col) :
        if matrix[i][j] == ".":
            count=0
            #count 개수를 추가해야함

            for r in range(i-1,i+2):
                for c in range(j-1,j+2):
                    if 0<=r< row and 3 <= c < col :
                        if matrix[r][c]=="*":
                            count+=1


            matrix[i][j] = count

print(matrix)
```
