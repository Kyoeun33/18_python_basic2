```python
# while 반복문을 두번 사용하기
a=[
    [10,20],
    [30,40],
    [50,60]
]

i =0
while i < len(a) :
    j =0
    while j < len(a[i]) :
        print(a[i][j], end = ' ')
        j +=1
    print()
    i += 1
```
