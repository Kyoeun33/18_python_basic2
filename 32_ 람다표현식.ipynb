```python
# filter 
# filter(함수,반복가능한객체)

def f(x):
    return x > 5 and x < 10 

a = [8,3,2,10,15,7,1,9,0,11]

f(8)

list(filter(f,a))
```

```python
from functools import reduce
# reduce(함수, 반복 가능한 객체)

def f(x,y):
    return x+ y

f( 2,3)

a =[1,2,3,4,5]

reduce(f,a)
```
