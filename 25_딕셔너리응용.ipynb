```python
# 딕셔너리 표현식 사용하기

# {키 : 값 for 키, 값 in dictionary}

keys = ['a','b','c','d']
imsi = dict.fromkeys(keys)

imsi

x = {key:value for key,value in imsi.items()}
print(x)
```

```python
# 딕셔너리 표현식에서 if조건문 사용하기

x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

for key, value in x.items() :
    if value == 20 :
        del  x[key]

print(x)
```

```python
x = {
    "a":10,
    "b":20,
    "c":30,
    "d":40,
}

y=x.copy()

for key,value in y.items() :
    if value == 20 :

        del x[key]

print(x)
```
