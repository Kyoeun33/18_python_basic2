```python
# # while 반복문 사용하기

# 초기식
# while 조건식 : 
#     반복할 코드
#     변화식

i = 0
while i < 100 :
    print('hello world')
    i += 1  # i=i+1
```

```python
#초기값을 1부터 시작하기

i=1
while i <= 100:
    print('hello world',i)
    i+=1
```

```python
for i in range(5) :
    for j in range(5) :
        if j <= i :
            print('@', end = ' ')
    print(" dk")
```

```python
# 초깃값을 감소시키기

i=5
while i>0:
    print('hello',i)
    i -=1   #i =i-1
    
```

```python
count = int(input())

i = 0
while i < count :
    print('hello',i)
    i+=1
    
```
