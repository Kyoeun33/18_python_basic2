```python
#회문판별하기
## 반복문으로 문자 검사하기

word = input("단어를 입력하세요")

is_palidrome = True
for  i in range(len(word)//2):
    if word[i]!=word[-1-i]:
        is_palidrome = False
      

print(is_palidrome)
```

```python
#시퀀스 뒤집기로 문자 검사하기
word = input()

print(word==word[::-1])
```

```python
# N-gram만들기
# 반복문으로 n- gram 출력하기
text = "hello"

for i in range(len(text)-1):
    print(text[i],text[i+1],sep="")
    
```

```python
text = "this is python script"

words = text.split()

for i in range(len(words)-1):
    print(words[i],words[i+1])
```

```python
# zip으로 2-gram 만들기

text= "hello"

two_gram= zip(text,text[1:])
list(two_gram)

for i in two_gram:
    print(i[0],i[1],sep="")
```
