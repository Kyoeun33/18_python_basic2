```python
# 중첩 루프 사용하기

for i in range(5):
    for j in range(5):
        print("j:",j,end=" ")
    
    print("i:",i,"\\n",sep="")
```

```python
# 사각형으로 별 출력하기

for i in range(5) :
    for j in range(5):
        print('*',end='')
    print()
```

```python
for _ in range(5) :   #안 쓰는 변수 _ 처리 가능
    for _ in range(5):
        print('*',end='')  # print는 기본적으로  end= '\n' 상태, 아무것도 지정하디 않아도 \n은 출력
    print()  
```

```python
# 사각형 모양 바꾸기

for i in range(3):
    for j in range(7):
        print('*',end='')
    print()
    
```

```python
#  계단식으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j <= i :
            print("*",end='')
    print()
```

```python
# 대각선으로 별 출력하기

for i in range(5) :
    for j in range(5) :
        if i == j :
            print("x", end='')
        else :
            print(" ",end='')
    print()
```

```python
num = int(input())

for i in range(num):
    
    print(' '*(num+1-i)+'*'*(2*i+1))

```

```python
for _ in range(5) :
    for _ in range(5) :
        if _ == _ :
            print("x", end='')
        else :
            print(" ",end='')
    print()
```
