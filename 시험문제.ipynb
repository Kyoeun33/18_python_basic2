```python
A = [int(x) for x in input().split()]

def analyze_scores(*args):
    count=0
    avg = sum(args)/len(args)
    for i in range (len(args)):
        if avg < A[i]:
            count += 1
    print('평균:',avg,'평균 이상인 수의 개수:',count)

analyze_scores(*A)
```

```python
inventory = {"노트북": 1500000, "마우스": 30000, "키보드": 50000, "모니터": 300000}

def checkout(**kwarg):
    total =0
    for kw,agr in (kwarg.items()):
        if kw in inventory:
            print(kw,"=",agr,'개:',agr*inventory[kw],'원')
            total = total + agr*inventory[kw]
        else :
            print("판매하지 않는 상품입니다:",kw)
    print('총 결제 금액:',total,'원')


checkout(모니터=2, 패드=1)
checkout(노트북=1, 마우스=2)
```
