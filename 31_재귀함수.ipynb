```python
#재귀호출에 종료 조건 만들기

def hello(count):
    if count == 0 :
        return

    print("hello world", count)

    count-= 1
    hello(count)
    
```

```python
hello(5)
```

```python
# 재귀호출 팩토리얼 구하기

def factorial(n):
    if n == 1 :
        return 1
    return n * factorial(n-1)
```

```python
print(factorial(5)) # 5*4*3*2*1


```

```python
# 31.4
def is_palindrome(word):
    if len(word)<=1:
        return True
    # 회문 판별 식
    if word[0] != word[-1]:
        return False
    else :
        word = word[1:-1]
        is_palindrome(word)

    


print(is_palindrome("hello"))
print(is_palindrome("level"))
```

```python
# 31.5
FA = [0,1]

def fib(x):
    if FA[-1] < x+2 :
        FA.append(FA[-1]+FA[-2])
        return fib(x)
    else :
        return FA[-1]

#######################
n= int(input())
print(fib(n))
```
